## Standardize Categories

```python
df["Category"]=df["Category"].str.strip().str.title()
```

```python
df["Category"].unique()
```

## Handling missing values

```python
df["Category"]=df["Category"].fillna(df["Category"].mode()[0])
```

```python
df["Price"]=pd.to_numeric(df["Price"],errors = "coerce")
```

```python
df["Price"].isnull().sum()
```

```python
df["Price"]=df["Price"].fillna(
    df.groupby("Category")["Price"].transform("mean")
)
```

```python
df.Price.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
df["Total"]=df["Total"].fillna(df["Quantity"]*df["Price"])
```

```python
df.isnull().sum()
```

## Remove duplicate rows

```python
df.duplicated().sum()
```

```python
df["Total"] = df["Quantity"] * df["Price"]
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates().copy()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

## Quantity correction 
Converted negative values to positive values 

```python
incorrect_total = df[df["Total"] != (df["Quantity"] * df["Price"])]
print(incorrect_total)
```

```python
df[df["Quantity"]<0]
```

```python
df["Quantity"] = df["Quantity"].abs()
```

## Fix wrong data types

```python
df.dtypes
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"],format="mixed",errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"])
df["Price"] = pd.to_numeric(df["Price"])
df["Total"] = pd.to_numeric(df["Total"])
```
